## 2. Matris Çarpımında Şekil Uyum Kuralı (Shape Matching)

> 📐 **Altın Kural:**
- İki matrisin çarpılabilmesi için **birinci matrisin sütun sayısı (iç boyut), ikinci matrisin satır sayısına eşit olmalıdır.**
- **İç Boyutlar Eşit Olmalı:** $(m \times \mathbf{n}) \cdot (\mathbf{n} \times p) \to (m \times p)$
- Sonuç matrisinin boyutu ise **dış boyutlar** $(m \times p)$ olacaktır.

## 4. Yapay Sinir Ağı Katmanı Mantığı (nn.Linear)

> 🧠 **Derin Öğrenmenin Temeli:**
- PyTorch'taki `nn.Linear(in_features, out_features)` katmanı perde arkasında sadece bir **Matris Çarpımı ve Vektör Toplaması** yapar:
$$Y = X \cdot W^T + b$$
- Burada $X$ Girdiler, $W$ Ağırlıklar (Weights) ve $b$ Yanlılıktır (Bias).
- Üzerine eklenen `torch.relu()` gibi aktivasyon fonksiyonları ise modele **doğrusal olmayan (non-linear)** öğrenme yeteneği katar.


```python
torch.manual_seed(42)

# 1. Example dataset: [study_hours, sleep_hours, social_media_hours]
# 3 students, each with 3 features

X = torch.tensor([
    [5.0, 7.0, 2.0],   # Student A: studies 5h, sleeps 7h, social 2h
    [3.0, 5.0, 5.0],   # Student B
    [8.0, 6.0, 1.0]    # Student C
])   # shape: (3, 3)

print("Input shape:", X.shape, "\n")

# 2. Linear model to predict the grade
# One output = predicted grade (with this data obviously we are not predicting anything right, it is just for demonstration)

model = torch.nn.Linear(in_features=3, out_features=1)

predicted_grades = model(X)
# (we add an activation function generally after these layers but in this case it will just return the same thing, no need to add it but if you want you can test it in step 4)
print("Predicted grades (from nn.Linear):\n", predicted_grades)
print("Output shape:", predicted_grades.shape, "\n")

# 3. What does the Linear layer REALLY does:
# grade = X @ W^T + b

W = model.weight        # shape: (1, 3)
b = model.bias          # shape: (1,)

manual_output = X @ W.T + b

print("Manual matrix multiplication output:\n", manual_output)
print("Matches nn.Linear output:", torch.allclose(predicted_grades, manual_output), "\n")

# 4. Add a simple "activation" 
relu_output = torch.relu(predicted_grades)
print("After ReLU (forcing positive values):\n", relu_output)
```

### 📌 03. Matris Çarpımı - Özet Cheat Sheet

| Operasyon | Sembol / Fonksiyon | Açıklama |
| :--- | :--- | :--- |
| **Element-wise Çarpım** | `A * B` | Karşılıklı elemanları tek tek çarpar |
| **Matris Çarpımı** | `A @ B` veya `torch.matmul(A, B)` | Satır $\times$ Sütun çarpımlarının toplamı |
| **Şekil Kuralı** | $(m \times \mathbf{n}) \cdot (\mathbf{n} \times p)$ | İç boyutlar $\mathbf{n}$ aynı olmalı! |
| **Transpoz** | `A.T` | Şekli tersine çevirir: $(m \times n) \to (n \times m)$ |
| **Linear Katman** | `nn.Linear(in, out)` | Matematiksel olarak: $Y = X \cdot W^T + b$ |
