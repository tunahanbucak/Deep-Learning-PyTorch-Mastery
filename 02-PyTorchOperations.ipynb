## 4. Zeros, Ones & Arange (Sıfır, Bir ve Aralık Tensörleri)

> 💡 **Özel Tensör Fonksiyonları:**
- `torch.zeros(size)`: Tümü sıfırlardan oluşan tensör (Maskeleme vb. işlemler için).
- `torch.ones(size)`: Tümü birlerden oluşan tensör.
- `torch.arange(start, end, step)`: Belirli aralıkta ardışık sayılar üretir.
- `torch.ones_like(input)`: Referans verilen tensör ile **aynı şekil ve veri tipinde** birlerden oluşan tensör üretir.


```python
zeros = torch.zeros(size =(3,2))
```

```python
zeros
```

```python
ones = torch.ones(size=(3,2))
```

```python
ones
```

```python
#aralıklar da oluşturabiliyoruz
vector_with_arange = torch.arange(start=0, end=10, step=1)
```

### 📌 02. Tensör Operasyonları - Özet Cheat Sheet

| Fonksiyon | Açıklama | Örnek |
| :--- | :--- | :--- |
| `torch.rand()` | 0 ile 1 arasında rastgele sayılar üretir | `torch.rand(size=(3, 3))` |
| `torch.zeros()` | Sıfır matrisi üretir | `torch.zeros(size=(2, 4))` |
| `torch.ones()` | Birler matrisi üretir | `torch.ones(size=(3, 2))` |
| `torch.arange()` | Belirli adım aralığında sayılar üretir | `torch.arange(start=0, end=10, step=2)` |
| `torch.ones_like()` | Referans tensörün şeklinde 1'ler üretir | `torch.ones_like(target_tensor)` |
| **Element-wise \*** | İki tensörün karşılıklı elemanlarını çarpar | `[1, 2] * [3, 4] \to [3, 8]` |
