## 4. Şekil Manipülasyonları

> 🛠️ **Temel Şekillendirme Araçları:**
- **`reshape(shape)`:** Tensör şeklini yeniden düzenler. Eleman sayısı toplamda aynı kalmalıdır.
- **`view(shape)`:** `reshape` gibidir ancak orijinal tensörle aynı bellek alanını (memory view) paylaşır. Birinde yapılan değişiklik diğerini de etkiler! (Yalnızca `is_contiguous()` olan tensörlerde çalışır).
- **`squeeze()`:** Boyutu `1` olan tüm gereksiz boyutları kaldırır (Örn: `(1, 9)` $\to$ `(9)`).
- **`unsqueeze(dim=N)`:** Belirtilen `N` indeksine `1` boyutunda yeni bir boyut ekler.
- **`stack([t1, t2], dim=N)`:** Tensörleri belirtilen boyut ekseninde üst üste yığar.
- **`permute(dims)`:** Boyut eksenlerinin sırasını değiştirir (Örn: `(224, 224, 3)` $\to$ `(3, 224, 224)`).

## 5. Çok Boyutlu İndeksleme ve Dilimleme (Indexing & Slicing)

> 🔍 **Dilimleme Kuralları:**
- Python listelerindeki `:` tüm elemanları seçme mantığı PyTorch'ta da aynen geçerlidir.
- Örn: `a[:, :, 0]` $\to$ 0. ve 1. boyutların tüm elemanlarını al, 2. boyutun sadece 0. indeksini getir.


```python
# indexing & slicing
```

```python
a = torch.arange(1,10,1)
```

```python
#: bütün değerleri al demek
```

```python
a[:,0,:]
```

```python
# 0. ve 1. boyutların tüm değerlerini alın, ancak 2. boyutun yalnızca 1. indeksini alın.
a[:, :, 1]
```

```python
# 0. boyutun tüm değerlerini alın, ancak 1. ve 2. boyutun yalnızca 1. indeks değerini alın.
a[:, 1, 1]
```

```python
# 0. ve 1. boyutun 0. indeksini ve 2. boyutun tüm değerlerini alın
a[0, 0, :] # same as a[0][0]
```

### 📌 04. Agregasyon & Manipülasyon - Özet Cheat Sheet

| Metot | Açıklama | Not / Örnek |
| :--- | :--- | :--- |
| `.mean()` | Ortalama alır | ⚠️ Yalnızca `float` tiplerde çalışır! (`t.float().mean()`) |
| `.argmax()` | En büyük elemanın indeksini döndürür | `tensor.argmax()` |
| `.reshape()` | Boyutu değiştirir | Eleman sayısı eşit olmalı |
| `.view()` | Aynı belleği paylaşarak boyutu değiştirir | `t.is_contiguous()` olmalı |
| `.squeeze()` | `1` genişliğindeki boyutları siler | `(1, 5, 1) \to (5)` |
| `.unsqueeze(dim=0)` | 0. indekse `1` boyutunu ekler | `(5) \to (1, 5)` |
| `.permute(2,0,1)` | Boyut sıralamasını değiştirir | `(H, W, C) \to (C, H, W)` |
